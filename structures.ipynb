1. Lists
Lists are used to store multiple items in a single variable

```python
fruits = ["apples", "banana", "cherry"]
print(fruits)
```

# Indexing/slicing of lists
This is to specify a characters or items


```python
fruits =("apples", "banana", "cherry", "melon", "oranges")
print(fruits[1:3])
print(fruits[2:])
```

# Lists are changable
- you can replace, add or manipulate an item/character in a list
Upon creating a temporary database; we use []


```python
fruits =["apples", "banana", "cherry", "melon", "oranges"]
fruits[1] = "lemons"
print(fruits)
```

```python
fruits =["apples", "banana", "cherry", "melon", "oranges"]
fruits.insert(3, "berry")
print(fruits)
```

```python
fruits =["apples", "banana", "cherry", "melon", "oranges"]
fruits.remove("apples")
print(fruits)
```

```python
fruits =["apples", "banana", "cherry", "melon", "oranges"]
fruits.append("mangoes")
print(fruits)
```

```python
fruits =["apples", "banana", "cherry", "melon", "oranges"]
fruits.append("mangoes")
print(fruits[0])
print(fruits[1])
print(fruits[3])
print(fruits[4])
print(fruits[5])
```

# Assignments sets and tuples

# Tuples
- Tuple items are ordered, unchangeable, and allow duplicate values.
- Tuples are unchangeable, meaning that we cannot change, add or remove items after the tuple has been created
- Tuples allow duplicate values
- store data in parenthesis


# Tuples are unchangable
- We cannot add,replace or manipulate items or characters
 

```python
mytuple = ("apple", "banana", "cherry")
print(type(mytuple))
```

```python
fruits = ("apples", "bananas", "cherry")
fruits[1] == "Melon"
print(fruits)
```

# Tuples slicicng/indexing


```python
fruits = ("apples", "bananas", "cherry")
print(fruits[1:])
print(fruits[:2])
```

3. Sets
- Sets are used to store multiple items in a single variable
- A set is a collection which is unordered, unchangeable*, and unindexed
- Sets are written with callibraces{}
- Does not allow duplication of items/characters i.e

```python
names = {"Sharon", "Peter", "Albert", "Faith"}
print(names)
```

```python
# Unordered
names = {"Sharon", "Peter", "Albert", "Faith"}
names.add("Fidel")
print(names)
```
