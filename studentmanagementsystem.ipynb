```python
books = ["Css", "Statistics", "Astrology", "Machines"]
borrowed = {}
USERNAME = "admin"
PASSWORD = "1234"
MAX_BOOKS = 2


def login():
    for _ in range(3):
        user = input("Username: ").strip()
        pwd = input("Password: ").strip()

        if user == USERNAME and pwd == PASSWORD:
            print("Login successful!")
            return True
        print("Wrong details!")

    print("Access denied!")
    return False


def find_book(name, collection):
    for b in collection:
        if b.lower() == name.strip().lower():
            return b
    return None


def borrow_book():
    student = input("Student name: ").strip()
    print("Available books:", books)
    book = find_book(input("Book to borrow: "), books)

    if book is None:
        print("Book unavailable.")
    elif len(borrowed.get(student, [])) >= MAX_BOOKS:
        print("Borrowing limit reached.")
    else:
        books.remove(book)
        borrowed.setdefault(student, []).append(book)
        print("Book borrowed successfully!")


def return_book():
    student = input("Student name: ").strip()
    book = find_book(input("Book to return: "), borrowed.get(student, []))

    if book is None:
        print("Book not found in student's records.")
    else:
        borrowed[student].remove(book)
        books.append(book)
        print("Book returned successfully!")


def view_borrowed():
    if not borrowed:
        print("No books currently borrowed.")
    for student, items in borrowed.items():
        print(f"{student}: {items}")


def main():
    if not login():
        return

    while True:
        print("\n1. View books\n2. Borrow book")
        print("3. Return book\n4. View borrowed\n5. Exit")
        choice = input("Choose: ").strip()

        if choice == "1":
            print("Available books:", books)
        elif choice == "2":
            borrow_book()
        elif choice == "3":
            return_book()
        elif choice == "4":
            view_borrowed()
        elif choice == "5":
            print("Goodbye!")
            break
        else:
            print("Invalid choice!")


if __name__ == "__main__":
    main()
```
