# Dicionaries
This is storing of data in key value pairs
Data is ordered, changeable and no duplication
Key, value pairs are seprated with a colon : (Key:Value)
Use of commas to terminate a line

```python
school = {
    "name" : "Eldohub", 
    "staff" : "Gift",
    "est" : 2019
}

print(school)

```

```python
# Accessing information using keys
school = {
    "name" : "Eldohub", 
    "staff" : "Gift",
    "est" : 2019
}

print(school["staff"])
print(school["est"])
```

```python
# Adding information
school = {
    "name" : "Eldohub", 
    "staff" : "Gift",
    "est" : 2019
}
school["student"] = "Peter"
print(school)
```

```python
# Duplicates
school = {
    "name" : "Eldohub", 
    "staff" : "Gift",
    "est" : 2019, #Did not duplicate
    "est" : 2023
}

print(school)
```

```python
# Duplicates
school = {
    "name" : "Eldohub", 
    "staff" : "Gift",
    "est" : 2019, #Did not duplicate
    "year" : 2023
}

print(school)
```
