```python
class Lion:
    def sound(self):
        print("Roar!")

class Goat:
    def sound(self):
        print("Maa!")

class Chicken:
    def sound(self):
        print("Cluck cluck!")

lion = Lion()
goat = Goat()
chicken = Chicken()

lion.sound()
goat.sound()
chicken.sound()
```

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2

rect = Rectangle(5, 3)
circle = Circle(4)

print("Rectangle area:", rect.area())
print("Circle area:", circle.area())
```

```python
class Person:
    def activity(self):
        print("I am a person")

class Student(Person):
    def activity(self):
        print("I am studying")

class Teacher(Person):
    def activity(self):
        print("I am teaching")

class Principal(Person):
    def activity(self):
        print("I am managing the school")

student = Student()
teacher = Teacher()
principal = Principal()

people = [student, teacher, principal]

for person in people:
    person.activity()
```

# Encapsulation 
This is about protecting data inside a class 
This involves keeping an objects data and the methods that work with data together in one class, while controlling how the data is accessed or modified

```python
# Without using encapsulation: creating methods of bank account
class BankSystem:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance
        
    def display_balance(self):
        print(f"Balance: Kshs {self.balance}")
        
account1 = BankSystem("Brian", 2000)

account1.display = 5000
account1.display_balance()

```

```python
class BankSystem:
    def __init__(self,name,balance):
        self.name = name
        self.__balance = balance # protecting the data inside the class
        
    def deposit(self,amount):
        if amount > 0:
            self.__balance += amount # self.__balance = self.__balance + amount
            print(f"The amount deposited is Kshs. {amount}")
        else:
            print("invalid deposit amount")
            
    def display_amount(self):
        print(f"Balance is Kshs. {self.__balance}")
        
account1 = BankSystem("Brian", 2000)

account1.deposit(5000)
account1.display_amount()
        
```

```python
# building a student registration system using encapsulation

class Student:
    def __init__(self, name, age, marks):
        self.name = name #Public identitfy
        self._age = age #Protected attribute
        self.__marks = marks # protecting the data inside the class
        
    def display_results(self):
        print(f"Student Name: {self.name}")
        print(f"Student Age: {self._age}")
        print(f"Student Marks: {self.__marks}")
        
    def update_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
            print(f"Marks updated to: {self.__marks}")
        else:
            print("Invalid marks...")

# User inputs
name = input("Enter student name: ")
age = int(input("Enter student age: "))
marks = float (input("Enter student marks:"))

student1 = Student(name, age, marks)
student1.update_marks(85)  # Example of updating marks
student1.display_results()
```
