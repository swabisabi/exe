# Polymorphism 
This is a concept that allows different objects to use the samee methods but performs different actions

```python
class dog:
    def sound(self):
        print("The dog barks loudly")
        
class cat:
    def sound(self):
        print("The cat meows")
        
class cow:
    def sound(self):
        print("The cow mows")
        
dog1 = dog()
cat1 = cat()
cow1 = cow()

#Visualize the output using methods
dog1.sound()
cat1.sound()
cow1.sound()
```

```python
# Using loops in polymorphism
class dog:
    def sound(self):
        print("The dog barks loudly")
        
class cat:
    def sound(self):
        print("The cat meows")
        
class cow:
    def sound(self):
        print("the cow mows")
        
#create objects from the class
animals = [dog(), cat(), cow()]

#use loops
for animal in animals:
    animal.sound()
```

Task: create several classes of vehicles with the method of move();

```python
class car:
    def move(self):
        print("The car drives on the road")
        
class plane:
    def move(self):
        print("The plane flies in the sky")
        
class bicycle:
    def move(self):
        print("The bicycle rides on the road")
        
#objects
vehicles = [car(), plane(), bicycle()]

for vehicle in vehicles:
    vehicle.move()
```

```python
#different vehicles with their parameters performing the same method but different actions
class car:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed
    def move(self):
        print(f"{self.brand} is driving at {self.speed} km/hr")
        
class plane:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed
    def move(self):
        print(f" {self.brand} is flying at {self.speed} knots")
        
class bicycle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed
    def move(self):
        print(f" {self.brand} is riding at {self.speed} km/hr")
        
vehicles = [car("Toyota", 80), plane("Boeing 747", 200), bicycle ("BMX", 50)] #create objects from class

for vehicle in vehicles:
    vehicle.move()
        
        
```

```python
#Imagine you are building a smart home system containig lights, fans and security alarms. Each device has its own name, and activate() method

class lights:
    def __init__(self, room):
        self.room = room
    def activate (self):
        print(f"lights in the {self.room} are ON")
        
class fan:
    def __init__(self, room):
        self.room = room
    def activate(self):
        print(f"please switch on the fan in the {self.room}")
        
class alarm:
    def __init__(self, room):
        self.room = room
    def activate(self):
        print(f"Remember to switch on the alarms in {self.room} upon fire outbreak")
        
smart_home = [lights("Bedroom"), fan("living room"), alarm("kitchen")]

for devices in smart_home:
    devices.activate()
```
