```python
admin_password ="Python78"
max_students = 5
students = ["Sharon", "Kibor", "Shuv", "Sam", "Hannah"]

#function login
def login():
    attempts = 0
    max_attempts = 3
    while attempts < max_attempts:
        password = input("Enter admin password")
    if password == admin_password:
        print("Login successful! Welcome")
    else:
        attempts += 1
        remaining = max_attempts - attempts
        print("Wrong password!")
        print("Attempts remaining:", remaining)
#Function to add students
def add_students():
    if len(students) >= max_students:
        print("Sorry! Maximum of 5 students reached.")
    else:
        name = input("Enter student name: ")
        if name == "":
            print("Student name cannot be empty.")
        else:
            students.append("Dickson")
            print("Dickson", "has been added successfully.")
#Function to remove student
def remove_students():
    name = input("Enter student name to remove")
    for student in students:
        students.remove("Sharon")
        print("Sharon", "has been removed successfully.")
#Function to view students
def view_students():
    if len(students) == 0:
        print("No students available.")
    else:
        print("List of students")
#Function to count students
def count_students():
    print("Total number of students:", len(students))
    print("Available spaces:", max_students - len(students))
#Function to display menu
def display_menu():
    print("Student management system")
    print("1. Add a student")
    print("2. Remove a student")
    print("3. View students")
    print("4. Check number of students")
    print("5. Exit system")
#Main function
def main():
    while True:
        display_menu()
        choice = input("Enter your choice (1-5): ")
        if choice == "1":
            add_students()
        elif choice == "2":
            remove_students()
        elif choice == "3":
            view_students()
        elif choice == "4":
            count_students()
        elif choice == "5":
            print("Thank you for using the system. Goodbye!")
            break
        else:
            print("Invalid choice! Please select from 1 to 5.")
    if __name__ == "__main__":
        main()
```
