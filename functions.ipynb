# Function
- def - Define a function
- greet - Name of the function
- () - placeholder for parameters/inputs
- : - start the function/introduce indentation 

```python
# Calling a function
def greet():
    print("Hello Everyone")
    print("Welcome to today's class")
greet() # Used to call the function name
```

```python
# Types of arguments
# 1.Positional argument - They depend on their position or order, by linking the parameter and the respective argument

def students(name, age):
    print(f"Student: {name}")
    print(f"Age: {age}")
    print()
students("Lazarus", 23)
students("Abraham", 24)
students("Sammy", 25)
```

```python
# Arbitrary arguments (*args)
# *args allows a function to receive any amount of positional arguments

#previously
def add(a,b,c):
    print(a+b+c)
add(10,20,30)
```

```python
# using *args
def add(*args):
    print(args)
add(10,20,30,40,70)
```

```python
def show_students(*args):
    for student in args:
        print(student)
show_students("Abel", "Mary", "Brian", "Tony")
    
```

```python
def vehicles(*cars):
    for car in cars:
        print(car)
vehicles("Bmw", "Amg", "Toyota")
    
```

```python
# Normal arguments and *args
def student(name, *subjects):
    print(f"Student_name: {name}")
    print("Subjects enrolled: ")

    for subject in subjects:
        print(subject)

student("Sammy", "Mathematics", "History", "Biology", "Computer")
```

```python
#Key words arguments(**kwargs)
def students(**school):
    print(school)
print("Welcome to python class")
students(name= "Alex", age =12, grade=8)
students(name= "Brian", age=10, grade=7, school= "Eldohub", location= "Langas")
students(name= "Ian", course= "Python", level= "Mid-level")
```

```python
def students(**school):
    for student in school:
        print(f"{student}: {school[student]}")
print("Welcome to python class")
students(name= "Alex", age =12, grade=8)
print()
students(name= "Brian", age=10, grade=7, school= "Eldohub", location= "Langas")
print()
students(name= "Ian", course= "Python", level= "Mid-level")
```

```python
#Shopping receipt
# Imagine a customer buys several items from a shop

def shopping_receipt(**shop):
    total = 0
    for item in shop:
        print(f"{item} : Kshs {shop[item]}")
        total = total + shop[item]
    print(f"Total amount: {total}")
shopping_receipt(bread = 70, Milk = 60, Sugar =120, Butter =200)
```

```python
#Create a function called shopping() that accepts different item prices and calculates the total cost
def shopping(**shop):
    total_cost = 0
    for item in shop:
        print(f"{item} : Kshs {shop[item]}")
        total_cost = total_cost + shop[item]
    print(f"total cost: {total_cost}")
shopping(meat = 90, toothpaste = 44, gum = 34)
```

```python
#Create a function called shopping() that accepts different item prices and calculates the total cost
def shopping(**items):
    total=0
    for price in items.values():
        total+=price
    print(f"Total cost : Ksh{total}")

shopping(gold = 40, lead = 80, milk = 80)
```

#Python Scope
- Local and global variables
- Global variable can be accessed from different parts of a program and is created outside the function
- Local variable is created inside a function

```python
# Global scope- exists outside the function

family_money = 5000
def show_money ():
    print(family_money)
show_money()
```

```python
# Local scope- exists within the function
def my_money():
    my_money= 300
    print(my_money)
my_money()
```
