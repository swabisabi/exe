Examples of operators are:

- Assignment operators
- Comparison operators
- Logical operators

Other types of operators are:

- Arithmetic operators 
These include use of addition, subtraction, division(/), multiplication(*), modulus(%), floor division(//), exponentiation(**)

```python
x = 300
y = 40
z = 60

a = 10
b = 4

print(x+y)
print(y+z)
print(z-y)
print(y-z)
print(x*y)
print(x/z) #retuns as a float
print(x//z) #reruns as an integer
print(x%y) #Modulus
print(a**b)
```

- comparison operators (Used to compare values)
Greater than(>), Less than(<), equal to(==), not equal to(!=), greater than or equal to(>=), less than or equal to(<=), 




```python
x = 300
y = 40
z = 60

print(x == y)
print(y != z)
print(x > y)
print(z < y)
print(x >= y)
print(x <= z)
```

- Logical operators
They are used to combine condition statements
a. AND; Returns True with both statements are correct i.e X <5 and X <10 given x =1
b. OR; Returns True if one of the statements is correct i.e x< 4 or x <7
c. NOT; Reverse the result; Returns False if the answer is correct



```python
x = 20
print(x <5 or x >10)
print(x <5 and x >10)
print(x >5 and x >10)


```
