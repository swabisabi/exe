# (self) parameter
this refers to the particular object that is being currently used 
It allows each object to store and access its own attributes and call its own methods


```python
class students:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    def display_results(self): #methods simply because we are calling the object of displaying results
        print(f"{self.name} has scored {self.score} percent.")
student1 = students("Faith", 70)
student2 = students("Abraham", 75)


student1.display_results() #Student1 is able to call his or her own results based on the methods defined using the self parameter
student2.display_results()
```

```python
# Create a student class with the attributes name, math ,english. Add a method called average_score()

class Students:
    def __init__(self, name, math, english):
        self.name = name
        self.math = math
        self.english = english 

    def average_score(self):
        print(f"{self.name} has an average score of {(self.math + self.english) / 2} marks in both subjects.")
student1 = Students("Faith", 70, 80)
student2 = Students("Abraham", 75, 85)# Positional arguments with the attributes of the objects
student3 = Students("Brian", 90, 88)

student1.average_score()
student2.average_score()
student3.average_score()
```

```python
# Create a bank system where customers can check balance, deposit and withdraw 

class BankAccount:
    def __init__(self, name, balance):
        self.name = name #store and visualize attributes
        self.balance = balance
    def check_balance(self): # first action or method to check the balance of the account
        print(f"{self.name} has an account balance of Kshs. {self.balance}")
    def deposit(self,amount):
        if amount > 0:
            self.balance += amount #self.balance = self.balance + amount deposited
            print(f"Kshs. {amount} deposited successsfully")
        else: 
            print("Invalid deposit amount")
    def withdraw(self, amount):
        if amount < 0:
            print("Invalid withdraw amount")
        elif amount > self.balance:
            print("Insufficient funds")
        else:
            self.balance -= amount #self.balance = self.balance - amount withdrawn
            print(f"Kshs. {amount} was withdrawn successfully")
            
account1 = BankAccount("Dennis", 500)
account2 = BankAccount("Faith", 1000)
account3 = BankAccount("Sharon", 1500)
account4 = BankAccount("Ian", 2000)

account1.deposit(1500)
account1.check_balance()     
print()
account2.withdraw(200)
account2.check_balance()       
            
        
    
```

```python
class Book:
    def __init__(self, title, author, price ):
        self.title = title
        self.author = author
        self.price = price
        
    def display_book(self):
        print(f"{self.title}, written by {self.author}, sold for Kshs. 320")

book1 = Book("The Firm", "John Grisham", 320)
book1.display_book()

```

```python
def turn_on(light):
    light["status"] = "on"
    print(f"The {light['room']} light is now ON")

def turn_off(light):
    light["status"] = "off"
    print(f"The {light['room']} light is now OFF")

light1 = {"room": "Bedroom", "status": "off"}
light2 = {"room": "Kitchen", "status": "off"}

turn_on(light1)
turn_on(light2)
turn_off(light1)
```

```python
light_status = "off"

def turn_on():
    global light_status
    light_status = "on"
    print("The light is now ON")

def turn_off():
    global light_status
    light_status = "off"
    print("The light is now OFF")

def check_status():
    print(f"The light is currently {light_status}")

check_status()
turn_on()
check_status()
turn_off()
check_status()
# why did we create variables inside the class and not global variables outside the class
```

# Changing the object properties 

```python
class student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade
        
#Creating objects from the class
student1 = student("Ian", 24, 10)
student2 = student("Kelly", 12, 7)
student3 = student("Bella", 9, 6)

# Changing the object properties
student1.age = 18
student2.name = "John"
student3.grade = 5

#Adding properties to the class
student1.school = "Stanford"
student2.school = "Bali"
student3.school = "Kolick"

# Displaying the output(visualization) of the object
print(student1.name, student1.age, student1.grade, student1.school)
print(student2.name, student2.age, student2.grade, student2.school)
print(student3.name, student3.age, student3.grade, student3.school)
```
