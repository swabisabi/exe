# Loops
This is simply repetition/iteration of a block of code. 
We have different types of loops; For loop & while loop

For loop - is a loop that repeats a block with a specific number of times 

```python
print("Hello World")
print("Hello World")
print("Hello World")
print("Hello World")
print("Hello World")
print("Hello World")
```

```python
# use of range()

for i in range(5):
    print("Hello World")
```

```python
for i in range(4):
    print(i)
# first loop = 0
# second loop = 1
# third loop = 2
# fourth loop = 3    
```

```python
for number in range(5):
    print(number)
```

```python
# Start and stop
for i in range(50, 69):
    print(i)

# Starts at 50, stops at 69 but does not include 69    
```

```python
# Countinf of steps, start, stop skip
# Selecting even numbers
for i in range(50, 69, 2):
    print(i)
```

```python
for i in range(50, 40, -1):
    print(i)
print("The end!!")
# the indented code is what gets repeated    
```

```python
for i in range(10, 0, -1):
    print(i)
```

```python
# Using for with if block
# Suppose we want to identify even numbers

for number in range(1, 11):
    if number % 2 == 0:
        print(number)
```

```python
# Using for in a list (looping through a list)
fruits = ["apples", "bananas", "oranges", "mangoes"]

for fruit in fruits:
    print(fruit)
    
```

```python
students =["Abu", "Sharo", "Lazarus", "Fai", "Sam"]

for student in students:
    print(student)
```

```python
# Using for loop with dictionaries
town = {
    "Shop 1" : {
        "name" : "Laptop",
        "price" : 50000,
        "quantity" : 5
    }, 
    "Shop 2" : {
        "name" : "Mouse",
        "price" : "1500",
        "quantity" : 10
    },
    "Shop 3": {
        "name" : "Keyboard",
        "price" : 2000,
        "quantity" : 7
    }
 }
#Introduce key, value pairs in a dictionary
#Represent the key = object, value = details and dictionary = town
#for key, value in dictionary.items():
for object, details in town.items():
    print(object)
    print("name:", details["name"])
    print("price:", details["price"])
    print("quantity:", details["quantity"])
    print()#skipping a line
```

# While loops
It is used to execute a set of statements as long as the condition is true
It iterates on infinite

```python
# To do an increment (i)
 
i = 1
while i < 6:
    print(i)
    i = i + 1
```

```python
count = 1

while count <= 16:
    print(count)
    count = count + 3
```

```python
# Countdown

count = 5

while count >= 1:
    print(count)
    count= count - 1
```

```python
password = ""
while password != "Python":
    password = input("Enter password: ")
print("Access Granted")
```

```python
#Guess a number

number = 0

while number != 10:
    number = int(input("enter the number: "))
print("correct number")    
```

```python
# Comining if-else with whilw loop

password = "Python123"
attempts = 0

while attempts < 5:
    insert_password = input("Enter password")

    if insert_password == password:
        print("Access Granted")
        break #Stop iteration/loop upon success
    else: 
        attempts = attempts + 1
        print(f"wrong password, try again, {5- attempts} attempts remaining")
if attempts == 5:
        print("Maximum attempts reached!. Access Denied")

```
