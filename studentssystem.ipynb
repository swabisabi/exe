```python
books = ["Css", "Statistics", "Astrology", "Machines"]
borrowed = {}
username = "admin"
password = "1234"


def login():
    for i in range(3):
        user = input("Username: ")
        pwd = input("Password: ")

        if user == username and pwd == password:
            print("Login successful!")
            return True

        print("Wrong details!")

    print("Access denied!")
    return False


def borrow_book():
    student = input("Student name: ")
    print("Available books:", books)
    book = input("Book to borrow: ")

    if book in books and len(borrowed.get(student, [])) < 2:
        books.remove(book)
        borrowed.setdefault(student, []).append(book)
        print("Book borrowed successfully!")
    else:
        print("Book unavailable or borrowing limit reached.")


def return_book():
    student = input("Student name: ")
    book = input("Book to return: ")

    if book in borrowed.get(student, []):
        borrowed[student].remove(book)
        books.append(book)
        print("Book returned successfully!")
    else:
        print("Book not found in student's records.")


def main():
    if login():
        while True:
            print("\n1. View books\n2. Borrow book")
            print("3. Return book\n4. Exit")
            choice = input("Choose: ")

            if choice == "1":
                print("Available books:", books)
            elif choice == "2":
                borrow_book()
            elif choice == "3":
                return_book()
            elif choice == "4":
                break
            else:
                print("Invalid choice!")

print()
main()
```
