# Error handling
This allows a python program to handle errors during execution without immediately crashing


```python
age = int(input("Enter age"))
print(f"I am {age} years old")
```

```python
try: 
    age = int(input("Enter your age"))
    print(f"You are {age} years old")
except ValueError:
    print("Invalid age. Please enter a number(int)")
```

```python
try:
    number1 = int(input("Enter first number"))
    number2 = int(input("Enter second number"))
    
    answer = number1/number2
    print(f"Answer: {answer}")
    
except ZeroDivisionError:
    print("A number cannot be divided by zero")
    
except ValueError:
    print("Insert a number(int), and also not a float")
    
```

try - executes a code that might raise an exception
except - handles a matching exception (fixing your own bugs)
else - runs if the block finishes without an exception
finally - runs when leaving the try statement, whether an exception occured or not


```python
try:
    marks = float(input("Enter the students marks: "))
    
except ValueError:
    print("Invalid marks ")

else:
    print(f"Marks entered : {marks}")
    
finally:
    print(f"Marks processed successfully: {marks}")
```

Simulate a temperature sensor. Ask the user to enter a temperature reading. Accept readings between -40°C and 125°C and classify the temperature as cold, normal, or hot

```python
while True:
    try:
        reading = input("Enter the temperature reading (or 'q' to quit): ")
        if reading.lower() == "q":
            break
        temperature = float(reading)
        if temperature >= 80:
            print("Hot")
        elif temperature <= 20:
            print("Cold")
        else:
            print("Normal")
    except ValueError:
        print("Invalid input")

```

```python
while True:
    try:
        reading = input("Enter the temperature reading in °C (or 'q' to quit): ")
        if reading.lower() == "q":
            break
        temperature = float(reading)

        if temperature < -40 or temperature > 125:
            print("Out of range: readings must be between -40°C and 125°C")
        elif temperature >= 80:
            print("Hot")
        elif temperature <= 20:
            print("Cold")
        else:
            print("Normal")
    except ValueError:
        print("Invalid input: please enter a number")
```
